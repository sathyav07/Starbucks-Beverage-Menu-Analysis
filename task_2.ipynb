```python
import pandas as pd


df = pd.read_csv('starbucks_drinkMenu_expanded.csv')

print("First 5 rows:")
display(df.head())

print("\nDataset Info:")
print(df.info())
```

```python
# Key columns summary statistics (Mean, Min, Max, Percentiles)
metrics = ['Calories', 'Sugars (g)', 'Total Fat (g)', 'Caffeine (mg)']
summary_stats = df[metrics].describe()

print("Descriptive Statistics for Key Metrics:")
display(summary_stats)
```

```python
# Convert 'Caffeine (mg)' and 'Total Fat (g)' to numeric, coercing errors
df['Caffeine (mg)'] = pd.to_numeric(df['Caffeine (mg)'], errors='coerce')
df['Total Fat (g)'] = pd.to_numeric(df['Total Fat (g)'], errors='coerce')

# Beverage category average calories
category_analysis = df.groupby('Beverage_category')[['Calories', 'Sugars (g)', 'Caffeine (mg)']].mean()


category_analysis_sorted = category_analysis.sort_values(by='Calories', ascending=False)

print("Category-wise Average Analysis:")
display(category_analysis_sorted)
```

```python

high_calorie_drinks = df[df['Calories'] > 300][['Beverage_category', 'Beverage', 'Beverage_prep', 'Calories', 'Sugars (g)']]

print(f"Total High-Calorie Drinks (>300 Calories): {len(high_calorie_drinks)}")
display(high_calorie_drinks.head(10))
```

```python
print(f"Overall Average Calories: {df['Calories'].mean():.2f}")
print(f"Maximum Calories in a drink: {df['Calories'].max()}")
print(f"Minimum Calories in a drink: {df['Calories'].min()}")
```
