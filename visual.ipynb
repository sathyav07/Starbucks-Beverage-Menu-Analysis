```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams.update({'figure.autolayout': True})


df = pd.read_excel('starbucks_drinkMenu_expanded.xlsx')
df = pd.read_excel('starbucks_drinkMenu_expanded.xlsx')

# --- Chart 1: Bar Chart - Category-wise Average Calories ---
plt.figure(figsize=(10, 6))
cat_calories = df.groupby('Beverage_category')['Calories'].mean().sort_values()
cat_calories.plot(kind='barh', color='seagreen')
plt.title('Average Calories by Beverage Category', fontsize=14, fontweight='bold')
plt.xlabel('Average Calories', fontsize=12)
plt.ylabel('Beverage Category', fontsize=12)
plt.savefig('category_calories_chart.png', dpi=300)
plt.show()
```

```python
# --- Chart 2: Scatter Plot - Calories vs Sugars (Relationship) ---
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='Sugars (g)', y='Calories', hue='Beverage_category', palette='tab10', s=100, alpha=0.8)
plt.title('Relationship between Sugars and Calories', fontsize=14, fontweight='bold')
plt.xlabel('Sugars (g)', fontsize=12)
plt.ylabel('Calories', fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.savefig('sugar_vs_calories_chart.png', dpi=300)
plt.show()
```

```python
# --- Chart 3: Box Plot - Distribution of Calories Across Categories ---
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Calories', y='Beverage_category', palette='Set2')
plt.title('Calorie Distribution Across Categories (Outlier Check)', fontsize=14, fontweight='bold')
plt.xlabel('Calories', fontsize=12)
plt.ylabel('Beverage Category', fontsize=12)
plt.savefig('calorie_boxplot.png', dpi=300)
plt.show()
```

```python
# --- Chart: Pie Chart - Proportion of Beverage Categories ---
plt.figure(figsize=(10, 8))

# Counting the number of drinks in each category
cat_counts = df['Beverage_category'].value_counts()

# Plotting pie chart with percentage labels
plt.pie(cat_counts, labels=cat_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette("Set2"))

plt.title('Proportion of Beverage Categories on Menu', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('category_pie_chart.png', dpi=300)
plt.show()
```

```python
# --- Chart: Pair Plot - Nutritional Variables Correlation ---
# Selecting key numeric nutritional columns for pairwise analysis
numeric_cols = ['Calories', 'Sugars (g)', 'Total Fat (g)', 'Caffeine (mg)']

# Generating pair plot using Seaborn with kernel density estimation on the diagonal
sns.pairplot(df[numeric_cols], diag_kind='kde', palette='tab10')

plt.suptitle('Pairwise Relationships of Nutritional Variables', y=1.02, fontsize=14, fontweight='bold')
plt.savefig('nutritional_pairplot.png', dpi=300)
plt.show()
```
