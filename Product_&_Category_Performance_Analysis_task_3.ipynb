```python
import pandas as pd

df = pd.read_excel('starbucks_drinkMenu_expanded.xlsx')
```

```python
df['Calories'] = pd.to_numeric(df['Calories'], errors='coerce')
df['Sugars (g)'] = pd.to_numeric(df['Sugars (g)'], errors='coerce')
df['Total Fat (g)'] = pd.to_numeric(df['Total Fat (g)'], errors='coerce')

category_performance = df.groupby('Beverage_category').agg(
    Total_Drinks=('Beverage', 'count'),
    Avg_Calories=('Calories', 'mean'),
    Avg_Sugar_g=('Sugars (g)', 'mean'),
    Avg_Fat_g=('Total Fat (g)', 'mean')
).sort_values(by='Avg_Calories', ascending=False)
```

```python
print("--- Category Performance Summary ---")
display(category_performance)
```

```python
df['Calorie_Level'] = df['Calories'].apply(lambda x: 'Low/Moderate Calorie' if x <= 200 else 'High Calorie')

calorie_segment = df['Calorie_Level'].value_counts()
print("\n--- Calorie Segment Breakdown ---")
print(calorie_segment)
```

```python
size_analysis = df.groupby('Beverage_prep')['Calories'].mean().sort_values(ascending=False)
print("\n--- Size (Beverage Prep) Wise Average Calories ---")
display(size_analysis)
```
